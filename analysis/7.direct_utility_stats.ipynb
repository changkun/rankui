```r
library(report)
library(lme4)
library(lmerTest)
library(BayesFactor)
library(tseries)
library(dplyr)
library(Kendall)
source('bayesMannWhitney.R')
```

# Text

```r
df <- read.csv('processed/direct_utility.csv')
df <- df[df$domain == 'text',]
df <- df[df$interface == 'pointwise',]
df <- df %>% group_by(userID, interface, iteration) %>% summarise(utility = mean(utility))
report(cor.test(df$iteration, df$utility, method='kendall', alternative='greater', exact=FALSE))
```

```r
df <- read.csv('processed/direct_utility.csv')
df <- df[df$domain == 'text',]
df <- df[df$interface == 'listwise-rating',]
df <- df %>% group_by(userID, interface, iteration) %>% summarise(utility = mean(utility))
report(cor.test(df$iteration, df$utility, method='kendall', alternative='greater', exact=FALSE))
```

```r
df <- read.csv('processed/direct_utility.csv')
df <- df[df$domain == 'text',]
df <- df[df$interface == 'listwise-ranking',]
df <- df %>% group_by(userID, interface, iteration) %>% summarise(utility = mean(utility))
report(cor.test(df$iteration, df$utility, method='kendall', alternative='greater', exact=FALSE))
```

```r
df <- read.csv('processed/direct_utility.csv')
df <- df[df$domain == 'text',]
df$interface <- factor(df$interface, levels=c('listwise-ranking', 'pointwise', 'listwise-rating'))

m1 <- lmer(utility ~ interface * iteration + (1|userID), data=df)
m2 <- lmer(utility ~ interface + (1|userID), data=df)
print(summary(m1), verbose=TRUE)
```
