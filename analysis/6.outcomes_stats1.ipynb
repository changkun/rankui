```r
library(report)
library(lme4)
library(lmerTest)
library(BayesFactor)
library(tseries)
library(dplyr)
library(Kendall)
source('bayesMannWhitney.R')
```

```r
df <- read.csv('processed/quality_progress_text.csv')

df$interface <- factor(df$interface, levels=c('listwise-ranking', 'pointwise', 'pairwise', 'listwise-rating', 'listwise-notie', 'listwise-cantie'))
df <- df %>% group_by(userID, interface, iteration) %>% summarise(length = mean(length))
mm <- lmer(length ~ interface * iteration + (1|userID), data=df)
summary(mm)
print(report(mm), verbose=TRUE)
```
