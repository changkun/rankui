```r
library(report)
library(lme4)
library(lmerTest)
library(BayesFactor)
library(tseries)
library(dplyr)
library(Kendall)
source('bayesMannWhitney.R')
```

```r
df <- read.csv('processed/quality_progress_image.csv')

df$interface <- factor(df$interface, levels=c('listwise-ranking', 'pointwise', 'pairwise', 'listwise-rating', 'listwise-notie', 'listwise-cantie'), ordered=FALSE)
df
```

```r
df <- df %>% group_by(userID, interface, iteration) %>% summarise(brightness = mean(brightness))
m <- lmer(brightness ~ interface * iteration + (1|userID), data=df)
summary(m)
print(report(m), verbose=TRUE)
```
